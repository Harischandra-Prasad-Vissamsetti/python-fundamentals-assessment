```python
# Question 1: Even or Odd. Write a Python program that takes an integer as input and determines whether the number is even or odd.
# A number is even if it divides by 2 with no remainder.

number= int(input("Enter a number:"))
if(number%2==0):     # % gives the remainder after division
  print(number,"is an even number")

else:
  print(number, "is an odd number")
```
