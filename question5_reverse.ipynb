```python
# Question 5: Reverse a String
# Write a Python program that accepts a string and prints its reverse.

text= input("Enter a string: ")

# Build the reverse by adding each character to the front
reversed_text = ""
for ch in text:
    reversed_text = ch + reversed_text

print("Reversed string:", reversed_text)
# Shortcut for reference: text[::-1] gives the same result
```
