```python
# Question 4: Count Vowels in a String
# Write a Python program that takes a string as input and counts the number of vowels (a, e, i, o, u) present in the string.

text= input("Enter a string: ")
vowels="aeiou"
count=0

# Check each character; lower() makes uppercase vowels count too
for ch in text.lower():
  if ch in vowels:
    count=count+1
print("Number of vowels is:", count)
```
